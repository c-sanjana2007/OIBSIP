- Mean

These statistical measures help us understand sales, quantity, discount and profit patterns in the dataset.

```python
# Display numerical columns
numeric_columns = df.select_dtypes(include=np.number).columns

print("Numerical Columns:")
print(numeric_columns)
```

### 3.2 Calculating Descriptive Statistics

We calculate the mean, median, mode and standard deviation for Sales, Quantity, Discount and Profit.

These measures help us understand the distribution and variation of the business data.

```python
# Select important numerical columns
data = df[['Sales', 'Quantity', 'Discount', 'Profit']]

# Calculate descriptive statistics
print("Mean:")
print(data.mean())

print("\nMedian:")
print(data.median())

print("\nMode:")
print(data.mode().iloc[0])

print("\nStandard Deviation:")
print(data.std())
```

### 3.4 Descriptive Statistical Observations

- The most frequent discount is 0.00.
- The average profit is 28.66, while the median profit is 8.67.
- Sales and profit show considerable variation, as indicated by their standard deviations.
- The difference between mean and median suggests that some high-value transactions influence the averages.

# 4. Sales Trend Analysis

Sales trend analysis helps us understand how sales performance changes over time.

In this section, we analyze monthly and quarterly sales trends using line charts to identify patterns and changes in sales performance.

### 4.1 Monthly Sales Analysis

We group sales data by month and calculate the total sales for each month to understand monthly sales performance.

```python
# Extract month and year from Order Date
df['Year_Month'] = df['Order Date'].dt.to_period('M')

# Calculate total monthly sales
monthly_sales = df.groupby('Year_Month')['Sales'].sum()

# Display monthly sales
print(monthly_sales)
```

```python
# Plot monthly sales trend
plt.figure(figsize=(12, 5))

monthly_sales.plot(kind='line', marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

### Monthly Sales Trend – Observations

- Monthly sales fluctuate considerably from 2014 to 2017.
- The highest monthly sales occurred towards the end of 2017.
- Some months recorded comparatively low sales.
- Overall, sales show an increasing trend towards 2017, despite monthly fluctuations.
- The variations indicate changes in sales performance over time.

### 4.2 Quarterly Sales Analysis

Quarterly sales analysis helps us understand sales performance across different quarters of each year.

We group the sales data by quarter and calculate total sales to identify quarterly patterns.

```python
# Extract year and quarter
df['Year_Quarter'] = df['Order Date'].dt.to_period('Q')

# Calculate total quarterly sales
quarterly_sales = df.groupby('Year_Quarter')['Sales'].sum()

# Display quarterly sales
print(quarterly_sales)
```

```python
# Plot quarterly sales trend
plt.figure(figsize=(12, 5))

quarterly_sales.plot(kind='line', marker='o')

plt.title('Quarterly Sales Trend')
plt.xlabel('Quarter')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

### Quarterly Sales Trend – Observations

- Quarterly sales fluctuate from 2014 to 2017.
- The highest quarterly sales were recorded in the last quarter of 2017.
- The lowest quarterly sales were recorded in the first quarter of 2015.
- The fourth quarter shows comparatively strong sales performance.
- Overall, quarterly sales indicate an increasing trend over the four-year period.

# 5. Product Performance Analysis

Product performance analysis helps identify popular products and understand which product categories generate the highest revenue.

In this section, we will:
- Identify the top 10 best-selling products.
- Analyze revenue generated by each product category.
- Visualize product performance using bar charts.

### 5.1 Top 10 Best-Selling Products

We calculate the total quantity sold for each product and identify the top 10 products with the highest sales quantity.

```python
# Calculate total quantity sold for each product
top_products = df.groupby('Product Name')['Quantity'].sum()

# Select top 10 best-selling products
top_10_products = top_products.sort_values(ascending=False).head(10)

# Display results
print(top_10_products)
```

### 5.2 Top 10 Best-Selling Products Visualization

A horizontal bar chart is used to visualize the top 10 products based on total quantity sold. It helps compare product popularity easily.

```python
# Create horizontal bar chart
plt.figure(figsize=(12, 6))

top_10_products.sort_values().plot(kind='barh', color='skyblue')

plt.title('Top 10 Best-Selling Products')
plt.xlabel('Total Quantity Sold')
plt.ylabel('Product Name')

plt.show()
```

### Top 10 Best-Selling Products – Observations

- Staples is the best-selling product with 215 units sold.
- Staple envelope ranks second with 170 units sold.
- Easy-staple paper ranks third with 150 units sold.
- Office stationery products appear frequently among the top-selling products.
- These products may require proper inventory planning to maintain availability.

### 5.4 Revenue Analysis by Product Category

We calculate the total revenue generated by each product category to identify which category contributes the most to overall sales.

```python
# Calculate total revenue by category
category_sales = df.groupby('Category')['Sales'].sum()

# Display category-wise revenue
print(category_sales)
```

```python
# Create bar chart for category-wise revenue
plt.figure(figsize=(9, 5))

category_sales.plot(kind='bar', color='orange')

plt.title('Revenue by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Revenue')
plt.xticks(rotation=0)

plt.show()
```

### 5.5 Revenue by Product Category – Observations

- Technology generates the highest revenue, approximately 836,000.
- Furniture generates approximately 742,000 in revenue.
- Office Supplies generates approximately 719,000 in revenue.
- Technology is the leading product category in terms of revenue.
- All three categories contribute significantly to overall sales.

# 6. Correlation Analysis

Correlation analysis helps us understand the relationship between numerical variables in a dataset.

In this section, we analyze the correlation between Sales, Quantity, Discount and Profit using a correlation matrix and heatmap.

### 6.1 Correlation Matrix

A correlation matrix shows the strength and direction of relationships between numerical variables.

Correlation values range from -1 to +1.

```python
# Select relevant numerical columns
correlation_data = df[['Sales', 'Quantity', 'Discount', 'Profit']]

# Calculate correlation matrix
correlation_matrix = correlation_data.corr()

# Display correlation matrix
print(correlation_matrix)
```

### 6.2 Correlation Heatmap

A heatmap represents correlation values using different colours. It makes it easier to identify positive, negative and weak relationships between variables.

```python
# Create correlation heatmap
plt.figure(figsize=(8, 5))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap')
plt.show()
```

### 6.3 Correlation Heatmap – Observations

- Sales and Profit show a moderate positive correlation of 0.48.
- Sales and Quantity show a weak positive correlation of 0.20.
- Discount and Profit show a weak negative correlation of -0.22.
- Sales and Discount have almost no linear correlation.
- Most numerical variables show weak relationships with each other.
- Correlation does not necessarily mean causation.

# 7. Additional Data Visualization

An additional visualization helps discover patterns and business insights beyond the basic sales analysis.

In this section, we analyze the relationship between discounts and average profit to understand how discount levels are associated with profitability.

### 7.1 Average Profit by Discount

We group the records by discount level and calculate the average profit for each discount level.

```python
# Calculate average profit for each discount level
discount_profit = df.groupby('Discount')['Profit'].mean()

# Display results
print(discount_profit)
```

```python
# Create bar chart
plt.figure(figsize=(10, 5))

discount_profit.plot(kind='bar', color='teal')

plt.title('Average Profit by Discount Level')
plt.xlabel('Discount')
plt.ylabel('Average Profit')
plt.xticks(rotation=0)
plt.axhline(0, color='black', linewidth=0.8)

plt.show()
```

### 7.3 Average Profit by Discount – Observations

- Lower discount levels of 0.00, 0.10, 0.15 and 0.20 show positive average profit.
- Higher discount levels generally show negative average profit.
- A discount of 0.50 records the lowest average profit.
- The results indicate that higher discounts are associated with lower profitability.
- Businesses should carefully evaluate discount strategies to maintain profitability.
